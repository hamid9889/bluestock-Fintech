live nav - sbi bluechip (119551)


```python
import requests

code = 119551
url = "https://api.mfapi.in/mf/" + str(code)
r = requests.get(url, timeout=60)
r.raise_for_status()
```

```python
import pandas as pd

nav = pd.DataFrame(r.json()["data"])
nav
```

```python
# write csv

nav.to_csv("../../../data/raw/nav_119551.csv", index=False)
print("saved nav_119551.csv, rows:", len(nav))
```
