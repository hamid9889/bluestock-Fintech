live nav - hdfc top100 direct (125497)


```python
import requests

code = 125497
url = "https://api.mfapi.in/mf/" + str(code)
r = requests.get(url, timeout=60)
r.raise_for_status()
```

```python
import pandas as pd

nav = pd.DataFrame(r.json()["data"])
nav
```

```python
# save to raw folder

nav.to_csv("../../../data/raw/nav_125497.csv", index=False)
print("saved nav_125497.csv, rows:", len(nav))
```
