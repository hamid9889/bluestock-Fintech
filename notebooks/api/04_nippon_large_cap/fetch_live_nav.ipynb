live nav - nippon large cap (118632)


```python
import requests

code = 118632
url = "https://api.mfapi.in/mf/" + str(code)
r = requests.get(url, timeout=60)
r.raise_for_status()
```

```python
import pandas as pd

nav = pd.DataFrame(r.json()["data"])
nav
```

```python
nav.to_csv("../../../data/raw/nav_118632.csv", index=False)
print("saved nav_118632.csv, rows:", len(nav))
```
