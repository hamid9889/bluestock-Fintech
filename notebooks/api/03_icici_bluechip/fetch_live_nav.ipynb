live nav - icici bluechip (120503)


```python
import requests

code = 120503
url = "https://api.mfapi.in/mf/" + str(code)
r = requests.get(url, timeout=60)
r.raise_for_status()
```

```python
import pandas as pd

nav = pd.DataFrame(r.json()["data"])
nav
```

```python
nav.to_csv("../../../data/raw/nav_120503.csv", index=False)
print("saved nav_120503.csv, rows:", len(nav))
```
