# Day 2: 08 Investor Transactions Cleaning
**Instructions:**
- Standardise transaction_type values (SIP/Lumpsum/Redemption)
- Validate amount > 0
- Fix date formats
- Check KYC status enum values

```python
import pandas as pd
import os

raw_path = r'../../data/csv_upload/08_investor_transactions.csv'
processed_dir = r'../../data/day2_processed/'

df = pd.read_csv(raw_path)
print(f"Initial rows: {len(df)}")
df.head()
```

```python
# 1. Fix date formats
df['transaction_date'] = pd.to_datetime(df['transaction_date'], dayfirst=True, errors='coerce')

# 2. Standardise transaction_type values
type_map = {
    'SIP': 'SIP',
    'Lumpsum': 'Lumpsum',
    'Redemption': 'Redemption',
    'STP': 'SIP', 
    'SWP': 'Redemption'
}
df['transaction_type'] = df['transaction_type'].map(lambda x: type_map.get(x, x))

# 3. Validate amount > 0
df = df[df['amount_inr'] > 0]

# 4. Check KYC status enum values
valid_kyc = ['Verified', 'Pending', 'Rejected']
df['kyc_status'] = df['kyc_status'].apply(lambda x: x if x in valid_kyc else 'Pending')

print(f"Final rows: {len(df)}")
```

```python
# Save to processed
df.to_csv(os.path.join(processed_dir, 'day2_investor_transactions_cleaned.csv'), index=False)
print("File saved to data/processed/investor_transactions_cleaned.csv")
```
