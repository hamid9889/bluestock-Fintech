# Day 2: API NAV Data Cleaning
Cleaning files fetched from API located in `data/raw/`.

```python
import pandas as pd
import os
import glob

raw_dir = r'../../data/raw/'
processed_dir = r'../../data/processed/'

api_files = glob.glob(os.path.join(raw_dir, 'nav_*.csv'))
print(f"Found {len(api_files)} API files")
```

```python
all_nav_data = []

for file in api_files:
    df = pd.read_csv(file)
    # Standardize columns if needed (assuming date, nav, amfi_code)
    df['date'] = pd.to_datetime(df['date'])
    all_nav_data.append(df)

df_combined = pd.concat(all_nav_data, ignore_index=True)
df_combined = df_combined.drop_duplicates(subset=['amfi_code', 'date'])
df_combined = df_combined.sort_values(['amfi_code', 'date'])

print(f"Combined API rows: {len(df_combined)}")
```

```python
# Save to processed
df_combined.to_csv(os.path.join(processed_dir, 'api_nav_combined_cleaned.csv'), index=False)
print("File saved to data/processed/api_nav_combined_cleaned.csv")
```
